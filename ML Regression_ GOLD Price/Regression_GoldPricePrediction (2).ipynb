```python
# Problem statement
           #predict pure gold rate based on year and 22k gold rate - Regression

# load the dataset

import pandas as pd


dataset=pd.read_csv("gold_rate_history.csv")

print(dataset.info())
print(dataset.head())
```

```python
# Data Preprocessing
       # 1. Missing values 2. duplicate entries

# 3. data type formatting

dataset.dtypes
```

```python
# convert Date Col to datetime format

dataset['Date']=pd.to_datetime(dataset['Date'])



# Extract useful time features

dataset['year']=dataset['Date'].dt.year
dataset['month']=dataset['Date'].dt.month
dataset['day']=dataset['Date'].dt.day
dataset['dayofweek']=dataset['Date'].dt.dayofweek
```

```python
dataset
```

```python
# Drop unnecssary columns

dataset=dataset.drop(['Date','Country','State','Location'],axis=1)
```

```python
dataset.to_csv('gold_features_target.csv',header=True)
```

```python
dataset
```

```python
# Define Features (X) and Target(Y)

features= ['year','month','day','dayofweek','Standard Gold (22 K)']
target=['Pure Gold (24 k)']

# Segregation of X and Y in dataset

X=dataset[features]
Y=dataset[target]
```

```python
# splitting the dataset into Train 80% and Testing set 20%

from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
# Select algorithm   5000 - 10 - error rate 50
 #                         - 100 -  error rate 1

from sklearn.ensemble import RandomForestRegressor

rf_model=RandomForestRegressor(n_estimators=100, random_state=42)

rf_model.fit(X_train,Y_train)
```

```python
# predict the gold price
y_pred=rf_model.predict(X_test)
```

```python
# Performance of the model

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Calculate performance metrics

mae=mean_absolute_error(Y_test,y_pred)
rmse=np.sqrt(mean_squared_error(Y_test,y_pred))
r2=r2_score(Y_test,y_pred)

# Print performance metrics

print("Mean Absolute Error:",mae)
print("Root Mean Squared Error:",rmse)
print("R-squared:",r2)
```

```python
import matplotlib.pyplot as plt

# plot actual vs predicted prices

plt.figure(figsize=(10,6))
plt.plot(Y_test.values, label='Actual', color='blue')
plt.plot(y_pred, label='Predicted', color='red',linestyle='dashed')
plt.xlabel('Sample')
plt.ylabel('Gold Price 24k')
plt.legend()
plt.title('Actual vs Predicted Gold Prices')
plt.show()
```

```python
# joblib, pickle  - purpose - serialization and deserialization

# serialization - object convet into byte streams - rf_model - file streams(pkl) - dump() - save the model into disk
# deserialization - reverse process of serialization - load() - import the model from disk

import joblib

# save or load your trained model
model = joblib.dump(rf_model, 'gold_price_prediction_model.pkl')
```
