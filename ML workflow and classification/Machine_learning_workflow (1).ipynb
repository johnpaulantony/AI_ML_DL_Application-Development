```python
# Data Segregation - X or features or input  , Y (Target var, label, output)

X=dataset.loc[:, "Age":"Salary"].values
X
```

```python
X=dataset.iloc[:,0:2].values
X
```

```python
Y=dataset.loc[:,'Status'].values
Y
```

```python
# Data Splitting - train and test data

# sklearn module - all ML alg written in this lib


!pip install scikit-learn

```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test=train_test_split(X,Y, test_size=0.2,random_state=0)
```

```python
len(X_train)
```

```python
len(Y_train)
```

```python
len(X_test) == len(Y_test)
```

```python
# feature scaling - model will give prioroty for all features during training

from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
X_train
```

```python
# select alg
from sklearn.linear_model import LogisticRegression
sales_model=LogisticRegression(random_state=0)
sales_model.fit(X_train,Y_train)
```

```python
from sklearn.metrics import accuracy_score
y_pred=sales_model.predict(X_test)
result_score=accuracy_score(Y_test,y_pred)
result_score
```

```python
age=int(input("enter the age"))
salary=int(input("enter the salary"))
new_user=sc.transform([[age,salary]])
new_user

result=sales_model.predict(new_user)
if result==1:
    print("customer will purchase")
else:
    print("customer will not purchase")
```
