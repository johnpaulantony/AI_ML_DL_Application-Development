```python
# import required libraries
from numpy import loadtxt # load data from csv file
from keras.models import Sequential #linear stack of layers in keras
from keras.layers import Dense  # fully connected neural network layer
from keras.models import model_from_json # Allows saving and loading a keras model in json format
```

```python
# loading the dataset
dataset=loadtxt('diabetes.csv',delimiter=',',skiprows=1)
```

```python
# splitting fatures (X)  and labels(Y)
X=dataset[:,0:8]
Y=dataset[:,8]
print(X)
print(Y)
```

```python
# Create the Neural Network
# Create the Neural Network
model=Sequential()
#adding layers
model.add(Dense(12,input_dim=8,activation='relu')) #98% relu function in hidden layer # input and hidden layer 1
model.add(Dense(8,activation='relu')) # hidden layer 2
model.add(Dense(1,activation='sigmoid')) #output layer
```

```python
# compile the model
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
# training the model
model.fit(X,Y,epochs=55,batch_size=10)
```

```python
# Evaluating the model
_,accuracy=model.evaluate(X,Y)
print('Accuracy: %.2f'%(accuracy*100))
```

```python
# save the model
model_json=model.to_json()
with open('model.json','w') as json_file:
    json_file.write(model_json)
model.save_weights('model.weights.h5')
print("saved model to disk")
```
