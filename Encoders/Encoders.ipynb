```python
'''Encoding Techniques
       - convert categorical data into numerical data  - Encoder

1. Label Encoding  - Inhertant order . Male, Female - natural order - Male - 1, Female - 0 , ( Order - High School, Bachelor, Master, Phd) - Tree based Models
2. One-Hot Encoding -
3. Ordinal Encoding
4. Target Guided Ordinal Encoding
5. Mean Encoding
6. Binary Encoding
7. Target guided MEan Encoding


'''
```

```python
# Label Encoding  - inhertant order . Tree-based   Avoid when Data is nominal

# Customer loyalty levels

from sklearn.preprocessing import LabelEncoder
import pandas as pd

# real world eg : customer loyalty levels

data =pd.DataFrame({'Loyalty':['Silver','Gold','Platinum','Silver','Gold']})
data

label_encoder = LabelEncoder()
data['Loyalty_encoder'] = label_encoder.fit_transform(data['Loyalty'])
data
```

```python
# ordinal Encoding - nominal order, Any Model  Avoid When - No order in data

from sklearn.preprocessing import OrdinalEncoder
import pandas as pd

data=pd.DataFrame({'Education':['High School','Bachelor','Master','PhD','Bachelor']})
data

# define order manually

education_order=['High School','Bachelor','Master','PhD']

ordinal_encoder = OrdinalEncoder(categories=[education_order])

# Fit and Transfrom
data['Education_encoder'] = ordinal_encoder.fit_transform(data[['Education']])
data


```

```python
# One Hot Encoding - Nominal Data , extend the features , Linear, Neural Networks, Avoid when - High Cardinality categorical fatures

from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# dataframe - payment method

data=pd.DataFrame({'Payment':['Credit Card','Debit Card','Paypal','Net Banking','Credit Card']})
data

onehot_encoder=OneHotEncoder(sparse_output=False)


# fit and transform

encoded_data=onehot_encoder.fit_transform(data[['Payment']])


# convert back to DataFrame
encoded_df=pd.DataFrame(encoded_data,columns=onehot_encoder.get_feature_names_out(['Payment']))

# concatenate

final_data=pd.concat([data,encoded_df],axis=1)
final_data
```

```python
# Frequency encoding - without extending the features,High Cardinality categorical fatures

import pandas as pd

# Product categories in an ecommernce platform - consider more than 100 unique features - don't use one hot encoding

data = pd.DataFrame({
    'product_category':['Electronics','Clothing','Electronics','Home','Clothing','Electronics','Beauty','Home']

})

# frequency encoding
freq_encoding = data['product_category'].value_counts()

# map the frequencies

data['product_Category_Encoded']=data['product_category'].map(freq_encoding)
data
```

```python
# problem statement - titanic survival prediction based on passenger details

# required lib
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,classification_report
import matplotlib.pyplot as plt
import seaborn as sns
import pickle
```

```python
# load the dataset
titanic_dataset=pd.read_csv('train.csv')
titanic_dataset
```

```python
# Data Preprocessing
# handle missing values

titanic_dataset.isna().sum()
```

```python
# fill the missing values
titanic_dataset['Embarked'].fillna(titanic_dataset['Embarked'].mode()[0])
```

```python
titanic_dataset['Age'].fillna(titanic_dataset['Age'].median(),inplace=True)
titanic_dataset['Embarked'].fillna(titanic_dataset['Embarked'].mode()[0],inplace=True)
titanic_dataset.drop(columns=['Cabin'],inplace=True)
```

```python
titanic_dataset.isna().sum()
```

```python
# segregate features and target  i.e X and Y

X=titanic_dataset[['Pclass','Sex','Age','SibSp','Parch','Embarked']]
Y=titanic_dataset['Survived']
```

```python
# Encoding in X

# Label Encoder

label_encoder=LabelEncoder()
X['Sex']=label_encoder.fit_transform(X['Sex'])
```

```python
X
```

```python
X=pd.get_dummies(X,columns=['Embarked'],prefix='Embarked')
```

```python
X
```

```python
# Spilt traing and test data

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
# train the model
model=LogisticRegression(max_iter=200)
model.fit(X_train,Y_train)
```

```python
# save the model

with open('titanic_model.pkl','wb') as model_file:
    pickle.dump(model,model_file)
```

```python
# load the model

with open('titanic_model.pkl','rb') as model_file:
    loaded_model=pickle.load(model_file)
#predict

sample_data=X_test.iloc[0:1]
Prediction=loaded_model.predict(sample_data)
Prediction
```

```python
# predict the result

accuracy_score=loaded_model.score(X_test,Y_test)
accuracy_score
```
